```python
import pandas as pd

data = {
    "Age": [18, 19, 20, 18, 21],
    "Study_Hours": [2, 5, 6, 1, 7],
    "Attendance": [70, 85, 90, 60, 95],
    "Result": ["Fail", "Pass", "Pass", "Fail", "Pass"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.head(2)
df.tail(2)
df.shape
df.info()
df.describe()
```

```python
plt.scatter(df["Study_Hours"], df["Attendance"])
plt.xlabel("Study Hours")
plt.ylabel("Attendance")
plt.title("Study Hours vs Attendance")
plt.show()
```
